# NLTK Setup and Word Tokenization

## Why Create an `nltk_data` Folder?

Installing `nltk` gives Python the NLTK library, but it does not install the language resources that many NLP tasks need. For example, word tokenization uses the `punkt` and `punkt_tab` resources.

This notebook creates an `nltk_data` folder in the project root. It keeps the downloaded resources in one predictable place so this notebook can find them every time it runs. The folder is created automatically if it does not already exist.

Run the next cell once after creating your project environment. It is also safe to run it again: NLTK skips resources that are already downloaded.

## Downloading the Resources

Each `nltk.download()` line downloads one resource and saves it in this project's `nltk_data` folder.

- `punkt` and `punkt_tab` are needed for word tokenization.
- `stopwords` provides common words such as `the`, `is`, and `and`.
- `wordnet` and `omw-1.4` are used for lemmatization.

`download_dir=str(NLTK_DATA_DIR)` selects the project folder as the download location. `quiet=True` hides detailed download messages, keeping the notebook output tidy.

## Basic Tokenization

`word_tokenize()` splits text into words and punctuation tokens.

````python
# Import Path so we can work with project folders safely.
from pathlib import Path

# Import the Natural Language Toolkit library.
import nltk

# Find the project folder because this notebook is inside notebook/.
PROJECT_ROOT = Path.cwd().resolve().parent
# Choose nltk_data in the project folder to store NLTK resources.
NLTK_DATA_DIR = PROJECT_ROOT / "nltk_data"
# Create the nltk_data folder when it does not already exist.
NLTK_DATA_DIR.mkdir(exist_ok=True)

# Tell NLTK to look in this project folder before other locations.
nltk.data.path.insert(0, str(NLTK_DATA_DIR))

# Download the basic sentence and word tokenizer resource.
nltk.download("punkt", download_dir=str(NLTK_DATA_DIR), quiet=True)
# Download the tokenizer tables required by recent NLTK versions.
nltk.download("punkt_tab", download_dir=str(NLTK_DATA_DIR), quiet=True)
# Download common words used when removing stopwords.
nltk.download("stopwords", download_dir=str(NLTK_DATA_DIR), quiet=True)
# Download WordNet, which is used for lemmatization.
nltk.download("wordnet", download_dir=str(NLTK_DATA_DIR), quiet=True)
# Download WordNet language data used with lemmatization.
nltk.download("omw-1.4", download_dir=str(NLTK_DATA_DIR), quiet=True)

# Show where the downloaded NLTK resources are stored.
print(f"NLTK data is ready in: {NLTK_DATA_DIR}")
````

````python
print(' word_tokenize:splits text into word and punctuation tokens.')
# Import the function that splits text into word and punctuation tokens.
from nltk.tokenize import word_tokenize # import word_ tokenize
from nltk.stem import WordNetLemmatizer  # import WordNetLemmantizer
import pandas as pd 
import random             # in order to select a random review
import spacy
import re
from sklearn.feature_extraction.text import CountVectorizer
from nltk.corpus import stopwords
# Store the sentence that we want to tokenize.
text = "I can't believe it's already Friday!"
# Split the sentence into a list of tokens.
tokens = word_tokenize(text)
# Print the list of tokens so we can inspect the result.
print('='*20)
print(tokens)
print('='*20)
print('sentence tokenization for splitting text into sentences:') 
print('='*20)
from nltk.tokenize import sent_tokenize

paragraph = "NLP is fascinating. It powers so many applications. I want to learn more."
sentences = sent_tokenize(paragraph)

print(sentences)
print('='*20)
print('Working with Stopwords in NLTK')
print('='*20)
from nltk.corpus import stopwords
print( 'List of Stopwords in NLTK:')
english_stops = set(stopwords.words('english'))
print(len(english_stops)) 

print('='*20)

print('Filtter stopwords from tokenized text:')

text = "The quick brown fox jumps over the lazy dog"
tokens = word_tokenize(text.lower())

filtered = [] #creates an empty list.

for word in tokens: #  checks each word in the token list one by one.
    if word not in english_stops: #  tests whether the word is not a stopword.
        filtered.append(word) # adds that word to the new list.

# or 
# filtered = [word for word in tokens if word not in english_stops]        
# the first word means:“Put this value into the new list.”
print(filtered)
````

## Understanding the spaCy Output

The output is valuable because spaCy gives information about every token in the sentence. Each output row has three values:

| Part | Meaning | Example |
| --- | --- | --- |
| `token.text` | The actual word or punctuation | `Apple` |
| `token.pos_` | The word's grammatical type | `PROPN` |
| `token.is_stop` | Whether spaCy considers it a stopword | `True` or `False` |

For example, this output:

```text
is AUX True
```

means:

- `is` is the word.
- `AUX` means auxiliary verb, or helping verb.
- `True` means spaCy considers `is` a stopword.

This information is useful for real NLP tasks. You can remove common words, select nouns and verbs, or identify important names in text.

## Disabling Pipeline Components
The full `spaCy pipeline` includes `parsing`, named entity recognition, and other components. If you only need tokenization, running the full pipeline wastes time.
You can disable components you don't need:

``` python
nlp = spacy.load('en_core_web_sm', disable=['parser', 'ner'])
```
*Neither approach is better. They serve different purposes. Use NLTK when learning concepts or when you need fine control. Use spaCy when building applications or processing large amounts of text.*

````python

nlp = spacy.load('en_core_web_sm', disable=['parser', 'ner'])
def clear_text(text):
    # Replace unwanted characters with SPACE (not empty string)
    text = re.sub(r"[^a-zA-Z']", " ", text)  # Note the space " " instead of ""
    text = text.split()
    text = " ".join(text) 
    
    return text
    

def lemmatize(text):
    doc = nlp(text.lower())
    lemmas = []
    for token in doc:
        lemmas.append(token.lemma_)
        
    return ' '.join(lemmas)

# store the review index in the review_idx variable
# either as a random number or a fixed value, e.g. 2557 
review_idx = 2557
# review_idx = 2557

review = corpus[review_idx]

print('The original text:', review)
print()
print('The lemmatized text:', lemmatize(clear_text(review)))
````

## Understanding the Body of `lemmatize()`

### 1. Process the text with spaCy

**`doc = nlp(text.lower())`**

Break this into two operations.

**`text.lower()`**

converts all letters to lowercase.

Then:

**`nlp(...)`** sends that lowercase text to the spaCy NLP model.

 `nlp ,as a pre-trained language-processing model,  analyzes text and provides information about words, such as their tokens, grammatical roles, and base forms (lemmas).`

```text
"The Dogs Are Running"
        ↓
text.lower()
        ↓
"the dogs are running"
        ↓
nlp()
        ↓
spaCy processed document
        ↓
doc
```
The processed result is stored in:
**`doc`**
So conceptually:
**`doc = processed version of the text`**
---
### 2. Create an empty list

**`lemmas = []`**
We need somewhere to collect the lemmas that spaCy finds.
At the beginning:
**`lemmas = []`**
It is empty.
Think of it as an empty container:
```text
lemmas
┌───────────────┐
│               │
└───────────────┘
```
---
### 3. Go through the text one token at a time
**`for token in doc:`**
`doc` contains the processed text.
spaCy breaks the text into **tokens**.
For example:
`"the dogs are running"`
might be treated as:
`the` → `dogs` → `are` → `running`
The loop says:
> For every token in `doc`, do the following operation.
So Python processes them one at a time:
```text
token = "the"
token = "dogs"
token = "are"
token = "running"
```
---
### 4. Find each token's lemma and save it

**`lemmas.append(token.lemma_)`**
There are two important pieces here.
**`token.lemma_`**
asks spaCy:
> What is the base form (lemma) of this token?
For example:
```text
dogs      → dog
are       → be
running   → run
```
Then:
**`.append()`**
means:
> Add this result to my `lemmas` list.
The list grows during the loop:
```text
Start:
lemmas = []

"the" → lemma → "the"
lemmas = ["the"]

"dogs" → lemma → "dog"
lemmas = ["the", "dog"]

"are" → lemma → "be"
lemmas = ["the", "dog", "be"]

"running" → lemma → "run"
lemmas = ["the", "dog", "be", "run"]
```

So this line:
**`lemmas.append(token.lemma_)`**
can be read as:

> Get the lemma of the current token and add it to the list.
---
### 5. Convert the list back into text
After the loop, we have something like:
**`["the", "dog", "be", "run"]`**
But we don't want to return a Python list. We want text.
Therefore:
**`' '.join(lemmas)`**
means:
> Join all the items in `lemmas` using one space between them.
It converts:
**`["the", "dog", "be", "run"]`**
into:
**`"the dog be run"`**
---
### 6. Return the result
**`return ' '.join(lemmas)`**
`return` sends the final result out of the function.
Therefore, the complete process is:
```text
INPUT
"The Dogs Are Running"
        ↓
text.lower()
        ↓
"the dogs are running"
        ↓
nlp()
        ↓
spaCy processes the text
        ↓
for token in doc
        ↓
process one token at a time
        ↓
token.lemma_
        ↓
find each base word
        ↓
lemmas.append(...)
        ↓
collect them in a list
        ↓
["the", "dog", "be", "run"]
        ↓
' '.join(lemmas)
        ↓
"the dog be run"
        ↓
RETURN
```
### How I should read the function
When I see:
**`doc = nlp(text.lower())`**
I read:
> Lowercase my text, process it with spaCy, and store the processed document in `doc`.
When I see:
**`lemmas = []`**
I read:
> Create an empty container for my results.
When I see:
**`for token in doc:`**
I read:
> Go through every token one at a time.
When I see:
``` python
lemmas.append(token.lemma_)
I read:
> Find the base form of this token and save it.
When I see:
**`return ' '.join(lemmas)`**
I read:
> Join all the lemmas with spaces and return the final text.
### Function pattern to recognize
This function follows a very common Python pattern:
```text
Prepare data
    ↓
Create empty container
    ↓
Loop through data
    ↓
Transform each item
    ↓
Store each result
    ↓
Combine results
    ↓
Return result
```
This pattern is more important to remember than memorizing the exact `lemmatize()` code.
